```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\SanyuPaul\Desktop\Final_Year_Project\CSV_Files\social_media_train.csv"
)

df.head()
```

```python
df["ff_ratio"] = df["#followers"] / (df["#follows"] + 1)
```

```python
df["log_followers"] = np.log1p(df["#followers"])
df["log_follows"] = np.log1p(df["#follows"])
```

```python
df["description length"] = df["description length"].clip(upper=300)
```

```python
features = [
    "profile pic",
    "nums/length username",
    "fullname words",
    "nums/length fullname",
    "name==username",
    "description length",
    "external URL",
    "private",
    "ff_ratio",
    "log_followers",
    "log_follows"
]

X = df[features]
y = df["fake"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)

class_weight_dict = dict(enumerate(class_weights))
class_weight_dict
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input

model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(64, activation="relu"),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
from sklearn.metrics import classification_report

y_pred = (model.predict(X_test) > 0.5).astype(int)
print(classification_report(y_test, y_pred))
```

```python
model.save("backend/models/fraud_model.h5")

import joblib
joblib.dump(scaler, "backend/models/scaler.pkl")
```

```python
# Get predicted probabilities (values between 0 and 1)
y_probs = model.predict(X_test)

# Just to see a few probabilities
y_probs[:10]
```

```python
y_pred_05 = (y_probs > 0.5).astype(int)

print("Threshold = 0.5")
print(classification_report(y_test, y_pred_05))
```

```python
y_pred_07 = (y_probs > 0.7).astype(int)

print("Threshold = 0.7")
print(classification_report(y_test, y_pred_07))
```

```python
y_pred_075 = (y_probs > 0.75).astype(int)

print("Threshold = 0.75")
print(classification_report(y_test, y_pred_075))
```

```python
def summarize(y_true, y_pred, label):
    from sklearn.metrics import accuracy_score
    print(f"{label} Accuracy:", accuracy_score(y_true, y_pred))

summarize(y_test, y_pred_05, "0.5")
summarize(y_test, y_pred_07, "0.7")
summarize(y_test, y_pred_075, "0.75")
```
